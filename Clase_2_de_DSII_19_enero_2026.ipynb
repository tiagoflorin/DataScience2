## 🧩 Definición del problema y obtención de datos

Antes de comenzar cualquier análisis de datos, es fundamental responder una serie de preguntas clave. Estas nos permiten orientar correctamente el trabajo y evitar esfuerzos innecesarios.

---

### ❓ ¿Qué problema debemos solucionar?

💡 **Explicación:**  
Antes de buscar datos, debemos tener completamente clara la pregunta que queremos responder.  
Si el problema no está bien definido, existe el riesgo de recolectar información irrelevante o difícil de interpretar.

📌 **Ejemplo:**  
> *“Queremos identificar cuáles son las provincias con mayor tasa de desempleo en Argentina durante 2024 para diseñar políticas públicas más efectivas.”*

---

### 📊 ¿Qué tipo de datos se requieren para hacer el análisis?

💡 **Explicación:**  
En esta etapa se determinan las variables necesarias para abordar el problema.  
Los datos pueden ser:

- **Cuantitativos:** valores numéricos (tasas, montos, cantidades).  
- **Cualitativos:** categorías o texto (provincia, sector, tipo de empleo).

📌 **Ejemplo:**  
Para el problema planteado, podríamos necesitar:

- Provincia  
- Tasa de desempleo  
- Población económicamente activa  
- Año o período de medición  

---

### 🌐 ¿Dónde podemos encontrar dichos datos?

💡 **Explicación:**  
Los datos pueden provenir de distintas fuentes, entre ellas:

- Organismos públicos  
- Bases de datos institucionales  
- APIs abiertas  
- Encuestas  
- Fuentes internacionales  

📌 **Ejemplo:**  
- El **INDEC** publica estadísticas oficiales de empleo.  
- El **Ministerio de Trabajo** ofrece información complementaria.  
- Organismos internacionales como el **Banco Mundial** disponen de bases de datos comparables entre países.

---

### 🔗 ¿Cómo podemos acceder a los datos?

💡 **Explicación:**  
Aquí se define el método de adquisición de la información, que puede variar según la fuente.

Algunas opciones comunes son:

- Descarga de archivos **CSV / Excel**  
- Consultas **SQL**  
- Consumo de **APIs**  
- **Web scraping**  
- Acceso a bases de datos internas  

📌 **Ejemplo:**  
- El INDEC permite descargar archivos en formato Excel desde su portal.  
- El Banco Mundial ofrece una **API** para acceder a los datos de forma automatizada.

---

### 🔍 ¿Los datos que deseamos realmente existen?

💡 **Explicación:**  
No siempre la información que buscamos está disponible en el nivel de detalle deseado.  
En esta etapa evaluamos si:

- Los datos existen  
- La frecuencia es adecuada (mensual, trimestral, anual)  
- El nivel de desagregación es suficiente  

Si no es así, puede ser necesario **ajustar el enfoque del análisis** o **buscar fuentes alternativas**.

📌 **Ejemplo:**  
Si se requieren datos **mensuales por provincia**, pero solo existen datos **trimestrales**, se deberá decidir si:

- Se adapta el análisis al período disponible  
- Se combinan fuentes  
- O se redefine el problema inicial  

---

✅ **Conclusión:**  
Responder estas preguntas antes de comenzar el análisis permite construir proyectos de Data Science más sólidos, coherentes y alineados con los objetivos del negocio o la política pública.

## 🔗 Cadena de Valor de Big Data

La **Cadena de Valor de Big Data** es un marco conceptual que describe las distintas fases por las que atraviesa la información, desde su obtención hasta su utilización final para generar **valor estratégico, operativo o analítico**.

Este modelo permite comprender que los datos **no generan valor por sí solos**, sino a través de un proceso continuo de transformación.

---

### 🧱 Etapas principales del proceso

La cadena se compone de **cinco grandes etapas**, cada una con un rol específico dentro del ciclo de vida de los datos.

---

## 1️⃣ Data Acquisition — *Adquisición de datos*

🔹 **Qué es:**  
Proceso de recopilación, filtrado y limpieza inicial de datos provenientes de múltiples fuentes, antes de su almacenamiento.

🔹 **Fuentes posibles:**  
- Datos estructurados (tablas, bases relacionales)  
- Datos no estructurados (texto, imágenes, audio)  
- Eventos digitales  
- Sensores IoT  
- Flujos de datos en tiempo real  

🔹 **Ejemplo:**  
- Descargar datos de sensores de tráfico  
- Extraer publicaciones de redes sociales  
- Recolectar registros de ventas de un sistema transaccional  

---

## 2️⃣ Data Analysis — *Análisis de datos*

🔹 **Qué es:**  
Transformación de los datos crudos en información útil que permita **comprender, explicar o predecir fenómenos**.

🔹 **Técnicas habituales:**  
- Minería de datos  
- Machine Learning  
- Análisis estadístico  
- Análisis semántico  
- Detección de patrones y tendencias  

🔹 **Ejemplo:**  
- Entrenar un modelo para predecir la demanda de un producto  
- Analizar el sentimiento de comentarios de clientes  

---

## 3️⃣ Data Curation — *Curación de datos*

🔹 **Qué es:**  
Gestión activa y continua de los datos a lo largo de su ciclo de vida para asegurar **calidad, consistencia y relevancia**.

🔹 **Incluye tareas como:**  
- Validación de datos  
- Eliminación de duplicados  
- Anotación y documentación  
- Control de calidad  
- Actualización periódica  
- Interoperabilidad entre sistemas  

🔹 **Ejemplo:**  
Revisar periódicamente una base de clientes para detectar registros duplicados, incompletos o desactualizados.

---

## 4️⃣ Data Storage — *Almacenamiento de datos*

🔹 **Qué es:**  
Almacenamiento de los datos en infraestructuras que permitan **acceso eficiente, seguridad y escalabilidad**.

🔹 **Opciones tecnológicas:**  
- Bases de datos en memoria  
- Sistemas NoSQL / NewSQL  
- Data Lakes  
- Almacenamiento en la nube  
- Sistemas distribuidos  

🔹 **Ejemplo:**  
- Hadoop o Spark  
- AWS S3, Azure Blob Storage, Google Cloud Storage  
- Bases de datos distribuidas  

---

## 5️⃣ Data Usage — *Uso de los datos*

🔹 **Qué es:**  
Aplicación práctica de los datos y del análisis realizado para **resolver problemas reales** o apoyar la toma de decisiones.

🔹 **Incluye:**  
- Predicciones  
- Visualizaciones  
- Dashboards  
- Modelos analíticos  
- Soporte a decisiones estratégicas  

🔹 **Ejemplo:**  
Un dashboard en Power BI que muestra el pronóstico de ventas y permite a un gerente optimizar inventarios.

---

### ✅ Idea clave

Este modelo permite comprender que:

> **La adquisición de datos es solo el inicio del proceso.**  
> El verdadero valor aparece cuando los datos atraviesan todas las etapas y se transforman en conocimiento accionable.

La Cadena de Valor de Big Data conecta la tecnología con el negocio y ayuda a entender cómo los datos pueden convertirse en decisiones informadas, innovación y ventaja competitiva.

## 📊 Tipos de datos según su estructura

En el análisis de datos es fundamental comprender **cómo están organizados los datos**, ya que esto determina la forma en que pueden almacenarse, procesarse y analizarse.

A continuación, se presentan los tres tipos principales de datos.

---

## 🧩 Datos estructurados

💡 **Idea clave:**  
Son como una **planilla de Excel perfectamente ordenada**.

- Se organizan en **filas y columnas**
- Cada columna tiene un significado claro
- Son fáciles de almacenar, consultar y analizar

🔹 **Características:**  
- Formato rígido  
- Estructura definida previamente  
- Alta eficiencia para búsquedas y consultas  

📌 **Ejemplo:**  
Una base de datos de clientes con los siguientes campos:

- Nombre  
- Apellido  
- Teléfono  
- Fecha de compra  

---

## 🧩 Datos semi-estructurados

💡 **Idea clave:**  
Tienen **cierto orden**, pero no siguen una estructura fija como una tabla.

- Utilizan **etiquetas o marcas** para identificar la información  
- El orden y la cantidad de campos pueden variar  

🔹 **Características:**  
- Estructura flexible  
- No dependen de filas y columnas  
- Muy comunes en aplicaciones web  

📌 **Ejemplo:**  
Un archivo **JSON** o **XML** que contiene información como:

```json
{
  "nombre": "Juan",
  "edad": 30,
  "ciudad": "Buenos Aires"
}

Lectura de datos con Pandas

📌 1. Lidiar con codificaciones raras (encoding)

Cuando abrimos un archivo y aparecen símbolos raros, Pandas permite indicar el tipo de codificación para leerlo correctamente.

Ejemplo: pd.read_csv("archivo.csv", encoding="latin1")

📌 2. Manipular encabezados y columnas

Podemos decidir qué fila es el encabezado o incluso cambiarle los nombres a las columnas.

Ejemplo: pd.read_csv("archivo.csv", header=2) (empieza desde la tercera fila)

📌 3. Trabajar con fechas

Pandas puede leer columnas como fechas y luego hacer operaciones con ellas (diferencias, orden, filtrado).

Ejemplo: pd.read_csv("ventas.csv", parse_dates=["fecha"])

📌 4. Definir tipos de datos desde el inicio

Podemos indicar si una columna es número, texto, fecha, etc., lo que ayuda a optimizar memoria y evitar errores.

Ejemplo: pd.read_csv("archivo.csv", dtype={"id_cliente": str})

📌 5. Identificar datos inválidos

Podemos encontrar valores nulos, duplicados o inconsistentes para luego limpiarlos.

Ejemplo: df.isnull().sum()

📌 6. Combinar y unir archivos

Pandas permite concatenar, unir y fusionar datasets fácilmente.

Ejemplo: pd.concat([df1, df2])

## 🌌 API de la NASA — APOD (Astronomy Picture of the Day)

La API que probaste corresponde a **APOD (Astronomy Picture of the Day)**.  
Su objetivo es brindar, en formato **JSON**, información sobre la imagen o el video astronómico destacado de un día específico.

---

### 📦 ¿Qué contiene la respuesta?

Cada consulta devuelve un objeto con los siguientes campos principales:

- **`date`** → Fecha del contenido astronómico  
  *(por ejemplo: `"2021-09-22"`)*

- **`title`** → Título de la imagen o video  
  *(ej.: “Equinox on a Spinning Earth”)*

- **`explanation`** → Descripción detallada escrita por astrónomos,  
  explicando el fenómeno observado o el contexto científico.

- **`media_type`** → Tipo de contenido:
  - `image`
  - `video`

- **`url`** → Enlace directo al recurso (imagen o video).

- **`service_version`** → Versión de la API utilizada.

---

### 💡 ¿Para qué sirve esta API?

La API APOD puede utilizarse para:

- 📘 Consultar **información educativa y científica** sobre el espacio.
- 🛰️ Obtener **imágenes y videos astronómicos de alta calidad**.
- 📊 Integrarla en **dashboards o aplicaciones** que muestren el contenido del día.
- 🎓 Usarla en **proyectos educativos** para practicar consumo de APIs y análisis de datos.

---

### 🔭 Ejemplo del resultado obtenido

En tu consulta, el resultado correspondía al **equinoccio**.  
La explicación describía cómo se observa desde un satélite la línea que separa el día y la noche sobre la Tierra, acompañada por un **video ilustrativo** que permite comprender visualmente el fenómeno.

# 🚀 ASTEROIDES

# Esto te da una tabla completa con todos los asteroides cercanos a la Tierra en un rango de fechas.

````python
import requests
import pandas as pd

url = "https://api.nasa.gov/neo/rest/v1/feed"
params = {
    "start_date": "2024-01-01",
    "end_date": "2024-01-05",
    "api_key": "ESKUwvKhYhaRXxKEcIRiCVrFbeU1KGaCSA16hada"
}

response = requests.get(url, params=params)
data = response.json()

# Convertir a DataFrame
asteroids = []
for date, objs in data["near_earth_objects"].items():
    for obj in objs:
        asteroids.append({
            "name": obj["name"],
            "date": date,
            "diameter_km": obj["estimated_diameter"]["kilometers"]["estimated_diameter_max"],
            "velocity_kph": obj["close_approach_data"][0]["relative_velocity"]["kilometers_per_hour"],
            "miss_distance_km": obj["close_approach_data"][0]["miss_distance"]["kilometers"]
        })

df = pd.DataFrame(asteroids)
print(df.tail())
````

````python
import pandas as pd

# Asegurar tipos numéricos
cols = ["miss_distance_km", "diameter_km", "velocity_kph"]
for c in cols:
    # Si vinieron con comas como separador de miles, las quitamos
    if df[c].dtype == "object":
        df[c] = df[c].astype(str).str.replace(",", "", regex=False)
    df[c] = pd.to_numeric(df[c], errors="coerce")

df_plot = df.dropna(subset=cols)

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sc = plt.scatter(
    df_plot["miss_distance_km"],
    df_plot["diameter_km"],
    c=df_plot["velocity_kph"]
)
plt.xlabel("Distancia de paso (km)")
plt.ylabel("Diámetro estimado (km)")
plt.title("Asteroides — Tamaño vs. Distancia (color = velocidad kph)")
cbar = plt.colorbar(sc)
cbar.set_label("Velocidad (kph)")
plt.tight_layout()
plt.show()
````

## 📊 Interpretación del gráfico de asteroides

Este gráfico permite analizar el **riesgo potencial de asteroides** combinando distancia, tamaño y velocidad.

---

### 🧭 Ejes del gráfico

#### ➡️ Eje X (horizontal): Distancia de paso (km)
Indica **qué tan cerca pasó el asteroide de la Tierra**.

- Valores cercanos a **0** → paso extremadamente cercano.
- Valores grandes (por ejemplo **7e7 = 70 millones de km**) → pasó mucho más lejos.

👉 Cuanto más a la derecha, **mayor distancia y menor riesgo inmediato**.

---

#### ⬆️ Eje Y (vertical): Diámetro estimado (km)
Representa el **tamaño máximo estimado del asteroide**.

- **0.1 km** → ~100 metros de diámetro  
- **1.0 km** → ~1 kilómetro de diámetro

👉 Cuanto más alto el punto, **mayor el tamaño del objeto**.

---

### 🎨 Color de los puntos: Velocidad de paso (km/h)

El color indica la **velocidad relativa del asteroide al pasar cerca de la Tierra**:

- 🟣 **Violetas** → velocidades bajas  
  *(≈ 15.000 – 30.000 km/h)*

- 🟢 **Verdes** → velocidades medias  
  *(≈ 40.000 – 70.000 km/h)*

- 🟡 **Amarillos** → velocidades muy altas  
  *(> 100.000 km/h)*

📌 La **barra de color a la derecha** funciona como leyenda que traduce cada color en velocidad.

---

### 🔍 ¿Cómo leer el gráfico?

- 🔹 **Punto bajo y a la izquierda (violeta)**  
  → asteroide pequeño, cercano y lento.

- ⚠️ **Punto alto y a la izquierda (amarillo)**  
  → asteroide grande, cercano y muy rápido  
  👉 *potencialmente más riesgoso*.

- ✅ **Punto bajo y a la derecha**  
  → objeto pequeño, lejano y lento  
  → riesgo muy bajo.

---

### 🚨 Conclusión

Los puntos **más altos y de color amarillo** son los que **requieren mayor monitoreo**, ya que combinan:

- gran tamaño  
- alta velocidad  
- cercanía relativa a la Tierra  

Estos son los principales candidatos para **vigilancia astronómica**.
