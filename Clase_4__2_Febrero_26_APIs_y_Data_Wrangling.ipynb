Definición sencilla

Una API (Application Programming Interface o Interfaz de Programación de Aplicaciones) es un conjunto de reglas que permite que dos sistemas o programas se comuniquen entre sí.


a mayoría de las APIs modernas usan peticiones HTTP (como cuando navegás por Internet).
El flujo básico es:

CLIENTE (tu programa) ---> solicita datos ---> API (servidor)
CLIENTE <--- recibe respuesta JSON <--- API


Request (solicitud) → Pedís información a la API (ej: el clima de Buenos Aires).

Response (respuesta) → La API devuelve la info, normalmente en formato JSON.

| **Tipo de API** | **Para qué sirve**                    | **Ejemplo real**         |
| --------------- | ------------------------------------- | ------------------------ |
| **REST**        | Solicita datos mediante URLs.         | Open-Meteo, Twitter API  |
| **GraphQL**     | Pide solo los datos necesarios.       | GitHub GraphQL API       |
| **Streaming**   | Envía datos en tiempo real.           | APIs de Trading o Twitch |
| **Locales**     | Llamadas dentro de tu propio sistema. | Windows API, Android API |

🍎 requests es una librería de Python que se usa para enviar solicitudes HTTP a servicios web o APIs. Es una de las herramientas más populares para conectarse con APIs y obtener datos desde internet.

https://thecatapi.com/

```python
import requests

# Paso 1: Definir la URL de la API
url = "https://api.thecatapi.com/v1/images/search"

# Paso 2: Hacer la solicitud GET
respuesta = requests.get(url)

# Paso 3: Verificar que la respuesta sea exitosa
if respuesta.status_code == 200:
    data = respuesta.json()
    image_url = data[0]['url']
    print("Imagen de gatito aleatoria:", image_url)
else:
    print("Error al obtener datos de la API:", respuesta.status_code)
```

* requests.get(url) → Envía una solicitud a la API.

* response.json() → Convierte la respuesta (en formato JSON) a una estructura que Python entiende.

* data[0]['url'] → Extraemos la URL de la imagen.

* Lo mostramos por pantalla.


🍎 explicacion codigo desde IF

* if response.status_code == 200:
👉 Comprueba si el servidor respondió correctamente.
El código 200 significa "OK" (todo salió bien).

* data = response.json()
👉 Convierte la respuesta (que viene en formato JSON) a una estructura de Python (una lista de diccionarios, en este caso).
Para convertir eso en un objeto de Python (como listas y diccionarios), y así poder trabajarlo fácilmente.

* image_url = data[0]['url']
👉 Accede al primer elemento de la lista (data[0]) y obtiene la URL de la imagen de gato desde la clave 'url'.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt

# 1. Hacer la solicitud
url = "http://books.toscrape.com/"
response = requests.get(url)
response.encoding = 'utf-8'  # 💥 esto resuelve el problema de 'Â'

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')

    # 2. Extraer datos
    libros = soup.find_all('article', class_='product_pod')

    titulos = [libro.h3.a['title'] for libro in libros]
    precios = [float(libro.find('p', class_='price_color').text.replace('£', '').strip()) for libro in libros]
    ratings_raw = [libro.find('p', class_='star-rating')['class'][1] for libro in libros]

    # 3. Convertir ratings a números
    rating_map = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
    ratings = [rating_map[r] for r in ratings_raw]

    # 4. Crear DataFrame
    df = pd.DataFrame({
        'Título': titulos,
        'Precio': precios,
        'Rating': ratings
    })

    print("📘 Primeros libros:")
    print(df.head())

    # 5. Gráfico de precios ordenados
    df_ordenado = df.sort_values(by='Precio', ascending=False)
    ax = df_ordenado.plot(kind='bar', x='Título', y='Precio', figsize=(12, 6), legend=False, color='skyblue')

    # Agregar etiquetas de precio
    for i, v in enumerate(df_ordenado['Precio']):
        ax.text(i, v + 0.5, f'£{v:.2f}', ha='center', va='bottom')

    plt.title("Precios de libros (ordenados)")
    plt.ylabel("Precio (£)")
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()

    # 6. Gráfico de dispersión: Precio vs Rating
    df.plot(kind='scatter', x='Rating', y='Precio', title="Relación entre Rating y Precio", figsize=(8, 5))
    plt.xlabel("Rating (estrellas)")
    plt.ylabel("Precio (£)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

else:
    print("❌ Error al acceder a la página:", response.status_code)
```

MCP estandariza cómo los modelos de lenguaje se comunican con herramientas externas (APIs, bases de datos, sistemas de archivos, etc.). En MCP, los conectores son los que hablan con las APIs y transforman la respuesta en un formato estándar.

```python
import requests

def get_weather(lat: float, lon: float):
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": lat,
        "longitude": lon,
        "current_weather": True
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = response.json()
        return {
            "temperature": data["current_weather"]["temperature"],
            "windspeed": data["current_weather"]["windspeed"],
            "status": "OK"
        }
    else:
        print("⚠️ Error:", response.status_code, response.text)
        return {"status": "FAIL"}

ciudades = {
    "Buenos Aires": (-34.6037, -58.3816),
    "Madrid": (40.4168, -3.7038),
    "New York": (40.7128, -74.0060)
}

for ciudad, coords in ciudades.items():
    result = get_weather(coords[0], coords[1])
    if result["status"] == "OK":
        print(f"🌎 {ciudad}: {result['temperature']}°C | 💨 {result['windspeed']} km/h")
    else:
        print(f"❌ Error consultando {ciudad}")
```

A. Uso tradicional de APIs

Usás get_weather() → llamás a Open-Meteo directamente.

Construís los parámetros manualmente.

Procesás la respuesta API → datos sin estandarizar.

Si mañana cambiás de API, tenés que reescribir código.

```python
import requests

# Simulamos un "conector MCP" para Open-Meteo
class MCPConnector:
    def __init__(self, api_url):
        self.api_url = api_url

    def query(self, latitude: float, longitude: float):
        """Consulta la API, pero devuelve la respuesta en formato MCP estandarizado"""
        params = {
            "latitude": latitude,
            "longitude": longitude,
            "current_weather": True
        }
        response = requests.get(self.api_url, params=params)
        if response.status_code == 200:
            data = response.json()
            return {
                "provider": "Open-Meteo",
                "location": {"lat": latitude, "lon": longitude},
                "data": {
                    "temperature": data["current_weather"]["temperature"],
                    "windspeed": data["current_weather"]["windspeed"]
                },
                "status": "OK"
            }
        else:
            return {
                "provider": "Open-Meteo",
                "location": {"lat": latitude, "lon": longitude},
                "status": "FAIL",
                "error": response.status_code
            }

# Creamos el conector MCP
mcp_weather = MCPConnector("https://api.open-meteo.com/v1/forecast")

# Definimos las ciudades
ciudades = {
    "Buenos Aires": (-34.6037, -58.3816),
    "Madrid": (40.4168, -3.7038),
    "New York": (40.7128, -74.0060)
}

# Usamos el "conector MCP" para hacer consultas
for ciudad, coords in ciudades.items():
    result = mcp_weather.query(coords[0], coords[1])
    if result["status"] == "OK":
        print(f"🌎 {ciudad}: {result['data']['temperature']}°C | 💨 {result['data']['windspeed']} km/h")
    else:
        print(f"❌ Error consultando {ciudad}")
```

Uso simulado de MCP

El modelo o agente no sabe nada de Open-Meteo.

Hay un conector MCP que:

Describe qué funciones están disponibles.

Estandariza cómo se devuelven los datos.

Si mañana cambiás Open-Meteo por WeatherAPI, el modelo sigue funcionando igual porque el formato MCP no cambia.

| **Característica**          | **API Directa**                          | **Con MCP**                                        |
| --------------------------- | ---------------------------------------- | -------------------------------------------------- |
| **Qué conoce el modelo**    | La URL, los parámetros, los campos JSON. | Solo sabe que existe una función `get_weather`.    |
| **Formato de la respuesta** | Depende de la API (puede cambiar).       | Estandarizado, uniforme.                           |
| **Cambio de proveedor**     | Tenés que reescribir código.             | No cambia el flujo.                                |
| **Casos de uso**            | Llamadas puntuales a una API.            | Integración de muchas APIs + agentes inteligentes. |

# 🔵 3 Paso - Hacer merge

```python
# ======================================
# Dimensión Fecha DERIVADA (OLTP)
# ======================================

# Copiamos fechas únicas desde la FACT
dates = (
    fact[["OrderDate"]]
    .drop_duplicates()
    .assign(
        year=lambda x: pd.to_datetime(x["OrderDate"]).dt.year,
        month=lambda x: pd.to_datetime(x["OrderDate"]).dt.month,
        month_name=lambda x: pd.to_datetime(x["OrderDate"]).dt.month_name(),
        quarter=lambda x: pd.to_datetime(x["OrderDate"]).dt.quarter,
        weekday=lambda x: pd.to_datetime(x["OrderDate"]).dt.weekday,
        weekday_name=lambda x: pd.to_datetime(x["OrderDate"]).dt.day_name()
    )
    .sort_values("OrderDate")
)

print("📅 Dimensión fecha derivada")
display(dates.head())
```

Es una tabla auxiliar que:

Parte de OrderDate

Genera features temporales

Se puede usar para:

análisis

agregaciones

modelos ML

Para análisis:

Para ML:

```python
# ======================================
# Merge FACT + dimensión fecha DERIVADA
# ======================================

ventas_fecha = fact.merge(
    dates,
    on="OrderDate",
    how="left",
    validate="many_to_one"
)

# ======================================
# Tipado de columnas
# ======================================
ventas_fecha["OrderDate"] = pd.to_datetime(
    ventas_fecha["OrderDate"],
    errors="coerce"
)

ventas_fecha["OrderQuantity"] = pd.to_numeric(
    ventas_fecha["OrderQuantity"],
    errors="coerce"
)

ventas_fecha["SalesAmount"] = pd.to_numeric(
    ventas_fecha["SalesAmount"],
    errors="coerce"
)

print("📊 Ventas + Fecha")
display(ventas_fecha.head())
```

```python
# ======================================
# Merge ventas + Customer (OLTP)
# ======================================

ventas_cli = ventas_fecha.merge(
    cust,
    on="CustomerID",
    how="left",
    validate="many_to_one"
)

print("👤 Ventas + Customer")
display(ventas_cli.head())
```

```python
# ======================================
# Merge ventas + Product (OLTP)
# ======================================

ventas_full = ventas_cli.merge(
    prod,
    on="ProductID",
    how="left",
    validate="many_to_one"
)

print("📦 Ventas + Customer + Product + Fecha")
display(ventas_full.head())
```

```python
# ======================================
# 2) Strings y nulos básicos
# ======================================
vf["FirstName"] = (
    vf["FirstName"]
    .str.strip()
    .str.title()
)

vf["LastName"] = (
    vf["LastName"]
    .str.strip()
    .str.title()
)

vf["EmailAddress"] = (
    vf["EmailAddress"]
    .str.strip()
    .str.lower()
)

vf["ProductName"] = (
    vf["ProductName"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

vf["Color"] = vf["Color"].fillna("Sin dato")
vf["Size"]  = vf["Size"].fillna("Sin dato")
```

```python
# ======================================
# 4) Derivadas útiles
# ======================================

vf["CustomerName"] = (
    vf["FirstName"].fillna("") + " " + vf["LastName"].fillna("")
).str.strip()

vf["Year"] = vf["OrderDate"].dt.year
```

🧠 Qué representa ventas_limpias

Una fila = una línea de venta

Claves + métricas + atributos de cliente y producto

Dataset listo para EDA y ML

```python
# ======================================
# 5) DataFrame final
# ======================================

ventas_limpias = vf[[
    "SalesOrderNumber",
    "CustomerID",
    "ProductID",
    "OrderDate",
    "OrderQuantity",
    "SalesAmount",
    "FirstName",
    "LastName",
    "EmailAddress",
    "ProductName",
    "Color",
    "Size"
]].reset_index(drop=True)
```

```python
# ======================================
# 6) Derivadas temporales y métricas
# ======================================

# Periodo mensual (YYYY-MM)
vf["YearMonth"] = vf["OrderDate"].dt.to_period("M")

# Ingreso unitario
vf["UnitRevenue"] = vf["SalesAmount"] / vf["OrderQuantity"]

# Revisión rápida
print(
    vf[[
        "OrderDate",
        "YearMonth",
        "SalesAmount",
        "OrderQuantity",
        "UnitRevenue"
    ]].head()
)
```

🔵 GroupBy
Ventas por año y género (Gender).

Ticket promedio por cliente (SalesAmount / OrderQuantity).

```python
# ======================================
# Ventas por año
# ======================================

ventas_anio = (
    vf.groupby("Year")["SalesAmount"]
      .sum()
      .reset_index()
      .sort_values("Year")
)

print("📊 Ventas por año:")
display(ventas_anio)
```

```python
# ======================================
# Segmentación: Empresa vs Persona
# ======================================

vf["CustomerType"] = vf["CompanyName"].apply(
    lambda x: "Empresa" if pd.notna(x) else "Persona"
)

ventas_anio_tipo = (
    vf.groupby(["Year", "CustomerType"])["SalesAmount"]
      .sum()
      .reset_index()
      .sort_values(["Year", "CustomerType"])
)

print("📊 Ventas por año y tipo de cliente:")
display(ventas_anio_tipo)
```

```python
# === Ticket promedio por cliente (IDs reales) ===

ticket_promedio_cliente = (
    ventas_limpias
    .groupby(
        ["CustomerID", "FirstName", "LastName", "EmailAddress"],
        as_index=False
    )
    .agg(
        SalesAmount_total=("SalesAmount", "sum"),
        OrderQuantity_total=("OrderQuantity", "sum")
    )
)

# Evitar división por cero
ticket_promedio_cliente = ticket_promedio_cliente[
    ticket_promedio_cliente["OrderQuantity_total"] > 0
]

# Ticket promedio
ticket_promedio_cliente["TicketPromedio"] = (
    ticket_promedio_cliente["SalesAmount_total"] /
    ticket_promedio_cliente["OrderQuantity_total"]
)

# Ordenar
ticket_promedio_cliente = ticket_promedio_cliente.sort_values(
    "TicketPromedio", ascending=False
)

print(" Top clientes por ticket promedio")
ticket_promedio_cliente.head()

```

```python
print(ventas_limpias.columns.tolist())
```

```python
# === APPLY: Clasificación por valor de cliente ===

def clasificar_valor(ticket):
    if pd.isna(ticket):
        return "Desconocido"
    elif ticket > 1000:
        return "Alto"
    elif ticket >= 500:
        return "Medio"
    else:
        return "Bajo"

# Usamos el DF de ticket promedio por cliente
ticket_promedio_cliente["NivelCliente"] = (
    ticket_promedio_cliente["TicketPromedio"]
    .apply(clasificar_valor)
)

print(ticket_promedio_cliente[[
    "CustomerID", "TicketPromedio", "NivelCliente"
]].head())
```

```python
ticket_promedio_cliente["EsPremium"] = (
    ticket_promedio_cliente["TicketPromedio"] > 1000
)

ticket_promedio_cliente.head()
```

```python
vf = vf.merge(
    ticket_promedio_cliente[["CustomerID", "NivelCliente", "EsPremium"]],
    on="CustomerID",
    how="left"
)
```

```python
# ================================
# EXPORTAR DATASET FINAL (vf)
# ================================

import pandas as pd
from google.colab import files

# --- Copia defensiva ---
vf_export = vf.copy()

# --- Asegurar fechas ---
if "OrderDate" in vf_export.columns:
    vf_export["OrderDate"] = pd.to_datetime(
        vf_export["OrderDate"], errors="coerce"
    ).dt.strftime("%Y-%m-%d")

# --- Period → string (YearMonth) ---
if "YearMonth" in vf_export.columns:
    if str(vf_export["YearMonth"].dtype).startswith("period"):
        vf_export["YearMonth"] = vf_export["YearMonth"].astype(str)

# --- Numéricos seguros ---
for col in ["OrderQuantity", "SalesAmount", "UnitRevenue"]:
    if col in vf_export.columns:
        vf_export[col] = pd.to_numeric(
            vf_export[col], errors="coerce"
        )

# --- Exportar CSV ---
output_file = "vf_ventas_limpias.csv"

vf_export.to_csv(
    output_file,
    index=False,
    sep=";",           # estándar LATAM / Excel
    decimal=",",
    encoding="utf-8-sig"
)

# --- Descargar ---
files.download(output_file)

print("✅ Dataset exportado correctamente:", output_file)
print("📐 Shape:", vf_export.shape)
vf_export.head()
```
